```python
import pandas as pd
import matplotlib.pyplot as plt

import numpy as np
```

```python
dataset =pd.read_csv("50_Startups_dataset.csv")
```

```python
dataset =pd.get_dummies(dataset,drop_first=True)
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
dependent = dataset[['Profit']]
                     
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.2,random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.svm import SVR
regressor=SVR(kernel="linear",C=10000)
regressor=regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test) 
y_pred
```

```python
from sklearn.metrics import r2_score 
percentage=r2_score(y_test,y_pred)*100
print(f"result: {percentage} %")
```

```python
import pickle 
filename = 'PaL_model.sav' 
pickle.dump(regressor, open(filename, 'wb'))
```
